## Objects, Types and Conversion

### 用 `type()` 看一个对象是什么类型

Python 里每个值都是对象，每个对象都有类型，类型决定它能做什么（能不能相加、能不能取下标）。
`type()` 把类型本身作为一个值返回，`print()` 会把它打印成 `<class '...'>` 的形式。
下面这五种类型是 Lecture 1 提到的标量类型：`int` 整数、`float` 实数、`bool` 真值、`NoneType` 空值、`str` 字符串。
读输出时注意最后一段：`"abc"` 有引号是字符串，`abc` 没有引号会被当成变量名而报错。

```python
print(5, type(5))          # 5 <class 'int'>
print(3.0, type(3.0))      # 3.0 <class 'float'>
print(True, type(True))    # True <class 'bool'>
print(None, type(None))    # None <class 'NoneType'>
print("abc", type("abc"))  # abc <class 'str'>
```

### 显式转换：`int()` 截断，`round()` 四舍五入

`int(x)` 的作用是**向 0 截断**：丢掉小数部分，不四舍五入，所以 `int(7.9)` 得到 7 而不是 8。
`round(x)` 才是四舍五入，`round(7.9)` 得到 8。负数能看出截断的方向：`int(-3.9)` 是 -3，向 0 靠拢而不是向下取到 -4。
最后一行演示 `round(2.5)` 与 `round(3.5)`：Python 在正好是 .5 时取最近的偶数，所以分别是 2 和 4，这一点和数学课的习惯不完全一样。

```python
print(int(7.9), round(7.9))    # 7 8
print(int(-3.9))               # -3
print(round(2.5), round(3.5))  # 2 4
```

### 隐式转换：`int` 与 `float` 混合运算

当一次运算里同时有 `int` 和 `float` 时，Python 会把 `int` 隐式升级为 `float`，结果也就是 `float`，这样可以保住小数部分。
`1 + 2.0` 得到 `3.0` 而不是 `3`，类型是 `<class 'float'>`。
这种自动转换只发生在"能安全表示"的类型之间：`"2" + 3` 不会自动把字符串变成数字，会直接报 `TypeError`，需要自己写 `int("2") + 3`。

```python
print(1 + 2.0, type(1 + 2.0))   # 3.0 <class 'float'>
print(1 + 2, type(1 + 2))       # 3 <class 'int'>
# print("2" + 3)                # 取消注释会报 TypeError
print(int("2") + 3)             # 5
```

## Expressions and Operators

### 四个容易混的运算符：`/` `//` `%` `**`

`/` 是除法，**结果一定是 `float`**，即使能整除：`6 / 3` 得到 `2.0`。
`//` 是向下取整除法（floor division），结果是"不超过真实商的最大整数"，操作数都是 `int` 时结果也是 `int`。
`%` 是取余（modulo），`**` 是幂。注意 `7 // 2` 与 `7 % 2` 合起来正好还原 `7 = 3 * 2 + 1`，这是后面写二进制转换时要用到的关系。

```python
print(7 / 2, type(7 / 2))      # 3.5 <class 'float'>
print(6 / 3, type(6 / 3))      # 2.0 <class 'float'>
print(7 // 2, 7 % 2)           # 3 1
print(2 ** 10)                 # 1024
```

### 运算顺序：先乘除后加减，`**` 最优先

没有括号时，Python 按 `**` → `* / // %` → `+ -` 的顺序计算，同级从左往右。
`13 - 4 / 12 * 12` 里先算 `4 / 12`，再乘 `12` 得到 `4.0`，最后 `13 - 4.0` 得到 `9.0`；即使结果看起来是整数，只要中间出现过 `/`，类型就是 `float`。
加上括号 `(13 - 4) / (12 * 12)` 就变成先减后除，得到 `0.0625`。看不准优先级时加括号永远是最稳的做法。

```python
print(13 - 4 / 12 * 12)        # 9.0
print((13 - 4) / (12 * 12))    # 0.0625
print(f"{(13 - 4) / (12 * 12):.4f}")   # 0.0625
```

## Variables and Binding

### 赋值是"先算右边，再绑定左边"

`=` 不是数学里的等号，而是一个动作：先求出右边的值，再把这个值绑定到左边的名字上。
所以 `pi = 355 / 113` 会先算出 `3.1415929203539825` 并存入内存，再让 `pi` 指向它。
左边必须是名字：`6 = x`、`x * y = 3 + 4` 都会报 SyntaxError。变量保存的是**值**，不是表达式本身。

```python
pi = 355 / 113
print(pi)              # 3.1415929203539825
radius = 2.2
circumference = pi * (radius * 2)
print(circumference)   # 13.823008849557525
```

### 重新绑定之后，已经算过的表达式不会自动更新

下面给 `radius` 换一个新值，变量 `area` 里存的仍然是**当时算出来的那个数**。
原因在第 3 行：`area = pi * (radius ** 2)` 只在那一行计算过一次，之后 `radius` 怎么变都不会牵动 `area`。
想让面积变成新的值，必须再写一次赋值。这是初学阶段最常见的逻辑错误之一（"我明明改了半径，面积怎么没变"）。

```python
pi = 3.14
radius = 2.0
area = pi * (radius ** 2)
print(area)             # 12.56

radius = radius + 1     # radius 变成 3.0
print(radius, area)     # 3.0 12.56（area 没有跟着变）
```

### 一个变量名在同一时刻只绑定一个值

`q = p` 做的是把 `p` 当前的值复制一份绑定给 `q`；随后给 `p` 赋新值，`q` 不受影响。
读这段代码时可以按"右边取值 → 左边绑定"两步在心里过一遍：先取 `p` 的值 5，再把它绑给 `q`，此时 `p` 和 `q` 各自指向 5。
最后 `p = 100` 只是让 `p` 改指向 100，`q` 依然指向 5。

```python
p = 5
q = p
p = 100
print(p, q)             # 100 5
```

### 交换两个变量：为什么第一种写法会失败

第一段想把 `x`、`y` 交换，但 `y = x` 的第一步就把 `y` 原来的值覆盖掉了，原来的 2 从此找不回来，所以第二句 `x = y` 只能拿到 1。
第二段用临时变量 `temp` 先把 `x` 的值存起来，再依次覆盖，最后把 `temp` 还给 `y`。
第三段是 Python 的一次性赋值 `c, d = d, c`：等号右边先整体求值（这时拿到的是两个旧值），再一起绑定，所以不需要临时变量。

```python
# 写法一：会失败
x, y = 1, 2
y = x       # y 变成 1，原来的 2 丢了
x = y       # x 也变成 1
print(x, y)             # 1 1

# 写法二：临时变量
a, b = 1, 2
temp = a
a = b
b = temp
print(a, b)             # 2 1

# 写法三：Python 的一次性赋值
c, d = 1, 2
c, d = d, c
print(c, d)             # 2 1
```

## Strings and Slicing

### 长度与索引：从 0 开始，负数从右边数

`len(s)` 返回字符个数。下标从 0 开始，所以长度为 11 的字符串，合法下标是 0 到 10。
`-1` 表示最后一个字符，`-2` 是倒数第二个，以此类推；`s[-1]` 与 `s[len(s) - 1]` 等价。
下标超出范围会报 `IndexError: string index out of range`，所以先看长度再谈下标。

```python
s = "abc d3f ghi"
print(len(s))            # 11
print(s[0], s[3], s[-1])  # a " " i（下标 3 是一个空格）
# print(s[11])           # 取消注释会报 IndexError
```

### 切片：`[start:stop:step]`，含头不含尾

切片取的是从 `start` 开始、到 `stop` **之前**停止的一段，所以 `s[3:6]` 取的是下标 3、4、5 三个字符。
三个数字都可以省略：省略 `start` 就从 0 开始，省略 `stop` 就取到结尾，省略 `step` 就是 1。
切片越界不会报错，只会把范围自动裁剪：`s[1:100]` 等价于取到结尾，`s[100:]` 得到空字符串。

```python
s = "abc d3f ghi"
print(s[3:6])            # " d3"
print(s[3:6:2])          # " d"（下标 3、5）
print(s[:3])             # abc
print(s[8:])             # ghi
print(s[1:100])          # bc d3f ghi
print(s[100:])           # 空字符串
print(s[6:3])            # 空字符串（start 比 stop 大，方向又没反）
```

### 步长为负：从右往左取

`step` 为负数时方向反过来：从 `start` 出发往左走，直到越过 `stop`。
`step = -1` 时最常用的写法是 `s[::-1]`：start 与 stop 都省略，等价于"从最后一个字符一路取到第一个"，也就是反转。
`s[4:1:-2]` 表示从下标 4 出发、往左每隔一个取一个，得到下标 4 和 2；注意此时**不含** stop 处的下标 1。

```python
s = "abc d3f ghi"
print(s[::-1])           # ihg f3d cba
print(s[4:1:-2])         # dc（下标 4、2）
print(s[-3:])            # ghi（最后三个字符）
```

### 拼接与重复：`+` 是拼接，`*` 是重复

`+` 把两个字符串接起来，两边都必须是字符串；`*` 让字符串重复若干次。
`"3" * 2 + "4"` 得到 `"334"`：先重复得到 `"33"`，再拼接 `"4"`。如果本意是算术，就必须先转类型，写成 `int("3") * 2`。
判断"这段代码在做算术还是在做字符串操作"，看操作数的类型就够了。

```python
print("3" * 2 + "4")     # 334
print(int("3") * 2)      # 6
print("ha" + "!" * 3)    # ha!!!
```

### 字符串不可变：只能造一个新字符串

`s[0] = "b"` 会报 `TypeError: 'str' object does not support item assignment`，因为字符串一旦创建就不能修改内部字符。
想要"改一个字符"，做法是用切片拼出一个新字符串：`"b" + word[1:len(word)]`，然后把变量重新指向新对象。
旧的 `"car"` 并没有被修改，只是没有名字再指向它了。理解这一点，后面看 `+=` 拼接字符串时才不会以为原串被改了。

```python
word = "car"
# word[0] = "b"          # 取消注释会报 TypeError
word = "b" + word[1:len(word)]
print(word)              # bar

text = "ab"
text = text + "c"        # 这是造了新字符串再重新绑定，不是修改原串
print(text)              # abc
```

## Input, Output and Formatting

### `input()` 读到的永远是字符串

`input("提示语")` 先把提示语打印出来，再读入一行，**返回值一定是字符串**，即使输入的是纯数字。
下面用变量 `text = "5"` 模拟用户输入：`5 * text` 是字符串重复，得到 `"55555"`；先 `int(text)` 再乘才得到 25。
写作业时"结果是一串重复的数字"或"类型不匹配报错"，绝大多数都是漏了这一步转换。

```python
text = "5"               # 模拟用户输入 5
print(5 * text)          # 55555（字符串重复）
print(5 * int(text))     # 25
# 真正的写法：text = input("Type a number: ")
```

### 一行读多个数：`split()` 与 `map()`

`input().split()` 把读到的一行按空白切分成若干段，返回值是一串字符串。
`map(int, ...)` 再把每一段都转成 `int`，配合左侧的多个变量名一次性赋值，写成 `a, b = map(int, input().split())`。
下面用 `line = "100 200"` 模拟这一行输入；如果输入用逗号分隔，就把 `split()` 写成 `split(",")`。

```python
line = "100 200"         # 模拟 input()
a, b = map(int, line.split())
print(a, b, a + b)       # 100 200 300
```

### 打印的三种写法

用逗号分隔多个对象时，`print` 会在每两项之间自动加一个空格，缺点是无法精细控制格式。
用 `+` 拼接时两边必须都是字符串，所以数字要写成 `str(...)`。
f-string 最省事：在字符串前加 `f`，`{}` 里可以写任意表达式，会在运行时求值并转成字符串；`{x:.2f}` 还能控制保留两位小数。
注意 `:.2f` 只改变显示形式，不会改变变量本身的值。

```python
num, fraction = 3000, 1 / 3
print(num * fraction, "is", fraction * 100, "% of", num)      # 1000.0 is 33.333333333333336 % of 3000
print(f"{num * fraction} is {fraction * 100:.2f}% of {num}")  # 1000.0 is 33.33% of 3000
w, h = 4, 3.8
print(f"面积 = {w * h:.2f}")                                  # 面积 = 15.20
```

## Branching

### 比较与逻辑运算的结果是 `bool`

`>` `<` `>=` `<=` `==` `!=` 这些比较运算，以及 `and` `or` `not` 这些逻辑运算，结果都是 `bool`，只有 `True` 和 `False` 两个值。
`and` 要求两边都为真才为真，`or` 只要一边为真就为真，`not` 取反。
下面用两层循环把所有组合列出来，读一遍这张表比死记要牢。

```python
for a in (True, False):
    for b in (True, False):
        print("a =", a, " b =", b, " a and b =", a and b, " a or b =", a or b, " not a =", not a)
```

### `if / elif / else`：只执行第一个条件为真的分支

程序从上往下检查条件，**一旦某个分支命中，就跳过整个 if 结构剩下的部分**，哪怕是后面还有成立的条件。
下面把 95、85、60、59 逐个转成等级：`score >= 90` 先判断，命中就是 A；不命中再看 `>= 80`，依此类推，最后 `else` 兜底。
判断顺序很重要：如果把 `>= 60` 写在最前面，95 分也会被判成 C。

```python
for score in (95, 85, 60, 59):
    if score >= 90:
        grade = "A"
    elif score >= 80:
        grade = "B"
    elif score >= 60:
        grade = "C"
    else:
        grade = "F"
    print(score, grade)
```

### 字符串比较和数值比较不是一回事

字符串之间用 `>` `<` 比较时按**字典序**逐字符比较，所以 `"10" < "9"` 是 `True`（先比首字符 `"1"` 与 `"9"`）。
要比数值大小，必须先把两边都转成数字：`int("10") < int("9")` 得到 `False`。
从 `input()` 拿到的数据都是字符串，这就是为什么读入数字后要立刻转换。

```python
print("10" < "9")              # True（字典序）
print(int("10") < int("9"))    # False
print("apple" < "banana")      # True
```

### 缩进决定一行属不属于某个分支

下面这段代码里，第 5、6 行都在 `if x == y:` 的缩进块内，第 11 行 `print("5: thanks")` 没有缩进，属于 `if` 结构之外，所以无论条件如何都会执行。
当 `x = y = 5` 时：`x == y` 成立进入第一层，`y != 0` 也成立，于是打印 1 和 2；`elif` 与 `else` 不再检查；最后打印 5。
输出的行号连起来是 125，这也是练习 8 要你判断的内容。

```python
x, y = 5, 5
if x == y:
    print("1: x and y are equal")
    if y != 0:
        print("2: x / y =", x / y)
elif x < y:
    print("3: x is smaller")
else:
    print("4: y is smaller")
print("5: thanks")
```

## Loops

### `for` 与 `range`：含头不含尾

`for i in range(...)` 每次循环让 `i` 取序列里的下一个值。`range(start, stop, step)` 生成从 `start` 开始、到 `stop` **之前**停止、间隔为 `step` 的整数。
只写一个数字时表示 `start = 0`、`step = 1`；写两个时 `step` 默认为 1；`step` 为负数就倒着数。
所以"从 1 数到 5"要写 `range(1, 6)`；`range(0)` 长度为 0，循环体一次都不执行。

```python
for i in range(5):
    print(i, end=" ")
print("   # range(5)")

for i in range(1, 4):
    print(i, end=" ")
print("   # range(1, 4)")

for i in range(1, 6, 2):
    print(i, end=" ")
print("   # range(1, 6, 2)")

for i in range(4, 0, -1):
    print(i, end=" ")
print("   # range(4, 0, -1)")

for i in range(0):
    print(i, end=" ")
print("   # range(0)")
```

### 累加器与计数器：初始化在循环外，更新在循环内

求和、计数、找最大值这类任务都用同一个套路：**循环开始前**把结果变量初始化，**循环体里**逐步更新它，循环结束后再使用。
下面第一段求 1 到 10 的和，注意 `mysum = 0` 必须在 `for` 之前，否则每轮都会被清零。
第二段统计字符串里元音字母的个数，`ch in "aeiou"` 判断当前字符是不是元音；`count` 同样在循环外初始化。

```python
mysum = 0
for i in range(1, 11):
    mysum += i
print(mysum)                   # 55

s = "demo loops - fruit loops"
count = 0
for ch in s:
    if ch in "aeiou":
        count += 1
print(count)                   # 8
```

### 阶乘：同一个问题用 `for` 和 `while` 各写一遍

`while` 版本要自己初始化 `i` 和 `factorial`，还要在循环体里让 `i` 递增，三件事缺一不可，否则会死循环或算错。
`for` 版本把"计数"交给 `range` 管理，只需要写乘法这一件真正要做的事，所以更短也更不容易出错。
两者结果相同（4! = 1×2×3×4 = 24）。选哪个的经验是：遍历已知的一串值用 `for`，重复次数不确定时用 `while`。

```python
x = 4

# while 版
i, factorial = 1, 1
while i <= x:
    factorial *= i
    i += 1
print(factorial)               # 24

# for 版
factorial = 1
for i in range(1, x + 1):
    factorial *= i
print(factorial)               # 24
```

### `break` 与 `continue`

`break` 立刻结束**当前这一层**循环，`continue` 只跳过本次循环剩下的语句，进入下一轮。
下面第一个循环里，`i == 3` 时 `continue` 跳过打印，`i == 6` 时 `break` 整体结束，所以输出是 1 2 4 5。
第二个循环是讲义里的例子：`mysum` 先加 5 变成 5，条件成立后 `break` 立即跳出，所以 `break` 下面那行 `mysum += 1` 永远不会执行，最终输出 5。

```python
for i in range(1, 8):
    if i == 3:
        continue
    if i == 6:
        break
    print(i, end=" ")
print()

mysum = 0
for i in range(5, 11, 2):
    mysum += i
    if mysum == 5:
        break
        mysum += 1      # 执行不到
print(mysum)            # 5
```

## Guess-and-Check

### 整数平方根：系统地一个个试，并想清楚什么时候停

思路是穷举枚举：从 `guess = 0` 开始，每次加 1，检查 `guess ** 2` 是否等于 `x`。
循环条件写成 `guess ** 2 < x`：一旦平方不再小于 `x` 就停下，因为再往上只会更大，没有继续试的必要。
循环结束后再判断 `guess ** 2 == x`：相等说明 `x` 是完全平方数，否则不是。这里用 `<` 而不是 `!=` 作循环条件，就是为了保证一定能停。

```python
for x in (16, 25, 24):
    guess = 0
    while guess ** 2 < x:
        guess = guess + 1
    if guess ** 2 == x:
        print("Square root of", x, "is", guess)
    else:
        print(x, "is not a perfect square")
# 输出：
# Square root of 16 is 4
# Square root of 25 is 5
# 24 is not a perfect square
```

### 边界情况：输入是负数时会怎样

当 `x = -2` 时，循环条件 `0 ** 2 < -2` 一开始就是假，循环体一次都不执行，`guess` 保持 0，于是走到 `else`，输出"不是完全平方数"。
这个结果不算错，但对用户没有帮助。更好的做法是先用一个布尔变量记下"输入是负数"这件事，输出时多给一句提示。
写循环时值得养成一个习惯：想一想输入取 0、负数、空串时会走到哪一步。

```python
x = -2
guess = 0
neg_flag = x < 0
while guess ** 2 < x:
    guess = guess + 1
if guess ** 2 == x:
    print("Square root of", x, "is", guess)
else:
    print(x, "is not a perfect square")
    if neg_flag:
        print("Just checking... did you mean", -x, "?")
# 输出：
# -2 is not a perfect square
# Just checking... did you mean 2 ?
```

### 布尔标志：记录循环里是否发生过某件事

问题是从 1 到 10 里找 `secret`，如果**没找到**要给出提示。麻烦在于循环结束后无法直接知道"到底找到没有"。
解决办法是立一个旗子：循环前 `found = False`，循环里一旦命中就改成 `True`，循环结束后检查它。
这种"布尔标志"（Boolean flag）的写法在查找、判断存在性时非常常用。

```python
secret = 7
found = False
for i in range(1, 11):
    if i == secret:
        print("yes, it's", i)
        found = True
if not found:
    print("not found")
# 输出：yes, it's 7
```

### 嵌套循环很慢：用已知条件减少循环层数

题目是 Alyssa、Ben、Cindy 卖票，三人共卖 N 张，Ben 比 Alyssa 少 2 张，Cindy 是 Alyssa 的 2 倍。
第一段用三层循环枚举所有组合，尝试次数是 (N+1)³：N = 38 时要试 59319 次。
第二段的做法是只枚举 Alyssa 一个人，Ben 和 Cindy 直接用式子算出来，尝试次数降到 39 次，结果完全相同。
题目规模变大时差距会非常夸张（N = 1000 时前者要试约 10 亿次），所以能用条件减少循环层数就不要暴力枚举。

```python
N = 38

slow = None
attempts = 0
for alyssa in range(N + 1):
    for ben in range(N + 1):
        for cindy in range(N + 1):
            attempts += 1
            if alyssa + ben + cindy == N and ben == alyssa - 2 and cindy == 2 * alyssa:
                slow = (alyssa, ben, cindy)
print("三层循环", attempts, "次 ->", slow)      # 59319 次 -> (10, 8, 20)

fast = None
attempts = 0
for alyssa in range(N + 1):
    ben = alyssa - 2
    cindy = alyssa * 2
    attempts += 1
    if ben + cindy + alyssa == N:
        fast = (alyssa, ben, cindy)
print("一层循环", attempts, "次 ->", fast)      # 39 次 -> (10, 8, 20)
```

## Binary and Floating Point

### 整数转二进制：`% 2` 取最低位，`// 2` 右移一位

把十进制整数写成二进制，本质是找一串 0/1 系数使 `x = a·2^n + ... + c·2^0`。
算法用的是两个已经学过的运算：`num % 2` 得到当前最低位，`num // 2` 把所有位整体右移一位，于是下一位变成新的最低位。
每一位要拼在结果的**前面**，因为先算出来的是最低位；整数除到 0 就结束。`19` 得到 `10011`，`1507` 得到 `10111100011`。

```python
for num in (19, 1507):
    original = num
    result = ""
    while num > 0:
        result = str(num % 2) + result
        num = num // 2
    print(original, "->", result)
# 输出：
# 19 -> 10011
# 1507 -> 10111100011
```

### 追踪一遍 19 的转换过程

为了看清每一步，这里把 `% 2`、`// 2` 和当时的结果都打印出来。
第一轮 `19 % 2 = 1`，`19 // 2 = 9`，结果拼成 `"1"`；第二轮取到 1，拼成 `"11"`；第三、四轮取到 0，拼成 `"011"`、`"0011"`；第五轮取到 1，得到 `"10011"`。
观察 `num` 的下降：19 → 9 → 4 → 2 → 1 → 0，每轮大约减半，所以循环次数约等于二进制位数。

```python
num = 19
result = ""
while num > 0:
    bit = num % 2
    print("num =", num, "  num % 2 =", bit, "  num // 2 =", num // 2, "  result =", str(bit) + result)
    result = str(bit) + result
    num = num // 2
print("19 ->", result)
```

### 二进制小数：0.125 能精确表示，0.1 不能

二进制小数的每一位对应 2 的负整数次幂：0.5、0.25、0.125、0.0625……
0.125 正好等于 1/8，也就是 2⁻³，可以用有限位精确表示；0.1 = 1/10 的分母含因子 5，二进制里是无限循环小数，只能存一个近似值。
第一段把 0.125 累加十次得到精确的 1.25；第二段把 0.1 累加十次得到 0.9999999999999999，可见误差会累积。
`0.1 + 0.2` 同样得到 0.30000000000000004，所以 `0.1 + 0.2 == 0.3` 是 `False`。

```python
y = 0
for i in range(10):
    y += 0.125
print(y, y == 1.25)            # 1.25 True

x = 0
for i in range(10):
    x += 0.1
print(x, x == 1)               # 0.9999999999999999 False
print(0.1 + 0.2)               # 0.30000000000000004
print(0.1 + 0.2 == 0.3)        # False
```

### 比较浮点数要用容差

既然浮点数只能近似表示，判断两个浮点数是否相等就不能用 `==`，而应判断它们的差是否足够小。
常见写法是 `abs(a - b) < 1e-9`：`1e-9` 是 10⁻⁹，可以根据需要的精度调整。
下面 `x` 实际是 0.9999999999999999，`x == 1` 为假，而 `abs(x - 1) < 1e-9` 为真，说明它们"在可接受的误差范围内相等"。

```python
x = 0
for i in range(10):
    x += 0.1
print(x == 1)                  # False
print(abs(x - 1))              # 1.1102230246251565e-16
print(abs(x - 1) < 1e-9)       # True
```

### 第 9 题：求 1 到 100 之间偶数的和

第一种写法用 `range(2, 101, 2)` 直接生成偶数序列，注意终点要写 101 才能包含 100（`range` 含头不含尾）。
第二种写法遍历 1 到 100，用 `i % 2 == 0` 筛出偶数再累加。

```python
total = 0
for i in range(2, 101, 2):
    total += i
print(total)                             # 2550

total = 0
for i in range(1, 101):
    if i % 2 == 0:
        total += i
print(total)                             # 2550
```

### 第 10 题：打印直角三角形

第一种写法每轮把 `"*" * i` 和一个换行拼到结果字符串后面，第 i 行就有 i 个星号。
第二种写法用嵌套循环：外层决定行数，内层用 `range(i)` 决定这一行拼几个星号。

### 第 11 题：统计第一个 z 之前的不同字母

用一个字符串变量 `seen` 记录已经见过的字母：遇到 `z` 立刻 `break`，遇到没见过的字母就拼进去。
最后 `len(seen)` 就是不同字母的个数。`"abzca"` 在 `z` 之前只有 `a`、`b` 两个不同字母，所以输出 2。
如果把 `break` 换成 `continue`，只是跳过 `z`，后面的 `c`、`a` 还会继续统计，结果会变成 3。

```python
s = "abzca"
seen = ""
for char in s:
    if char == "z":
        break
    if char not in seen:
        seen += char
print(len(seen))                         # 2
```

### 第 12 题：1507 的二进制表示

用前面讲过的算法自己算一遍：`% 2` 取最低位拼到最前面，`// 2` 右移一位，直到 `num` 变成 0。
Python 也提供了内置函数 `bin()` 可以直接得到结果，它返回的字符串带 `0b` 前缀，用 `[2:]` 去掉即可。
两种方式都得到 `10111100011`。

```python
num = 1507
result = ""
while num > 0:
    result = str(num % 2) + result
    num = num // 2
print(result)                            # 10111100011

print(bin(1507))                         # 0b10111100011
print(bin(1507)[2:])                     # 10111100011
```

## Common Pitfalls

写作业前扫一眼这张表即可。表里的现象在前面各节都出现过，这里只做汇总，不再重复讲解。

| 易错点 | 正确做法 |
|:---|:---|
| `input()` 直接拿来做算术 | `int(input())` / `float(input())` |
| `=` 和 `==` 混用 | 赋值用 `=`，比较用 `==` |
| 缩进层级写错 | 缩进决定代码块，统一 4 个空格 |
| `range(a, b)` 当成了含 b | 含头**不含尾**，要含 b 就写 `b + 1` |
| `int(7.9)` 以为是 8 | 那是截断；四舍五入用 `round()` |
| 用 `==` 比较浮点数 | 用 `abs(a - b) < 1e-9` |
| 切片越界以为会报错 | 切片只会给空串 `''`，索引越界才报错 |
| 循环里忘了更新变量 | 检查"条件什么时候变 False" |
| `break` 以为能跳出所有循环 | 只跳出**最内层**那一层 |
| 字符串和数字直接比大小 | 先转换类型 |
| 嵌套循环暴力枚举 | 先用条件减少循环层数 |
| 变量名随手起 `a`、`tmp` | 名字要说明含义 |

最后补两条前面没有单独演示的规则：`"3" * 2` 是字符串重复而不是乘法；`break` 只结束最内层循环，外层循环会继续。
下面第二个循环里，内层 `j == 1` 时跳出内层，外层 `i` 继续，所以输出是 `0 0  1 0  2 0`。

```python
print("3" * 2 + "4")     # 334
print(int("3") * 2)      # 6

for i in range(3):
    for j in range(3):
        if j == 1:
            break           # 只跳出内层
        print(i, j, end="  ")
print()                      # 0 0  1 0  2 0
```
