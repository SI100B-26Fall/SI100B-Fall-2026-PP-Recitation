## Lists

### 列表的索引与切片

```python
hello_list = [2, "a", 4, [1, 2]]
print(hello_list[0])      # 2
print(hello_list[1:3])    # ["a", 4]
print(len(hello_list))    # 4
print([1, 2] + [3, 4])    # [1, 2, 3, 4]
print([1, 2] * 2)         # [1, 2, 1, 2]
print(max([3, 5, 0]))     # 5
```

```python
L = [2, 4, 3]
L[1] = 5
print(L)      # [2, 5, 3]，改的是 L 自己，没有新建列表
```

### append 的返回值是 None

```python
L = [2, 1, 3]
print(L.append(5))   # None —— append 只改列表，不返回值
print(L)             # [2, 1, 3, 5]

L = [2, 1, 3]
L = L.append(5)      # 把 None 赋给了 L
print(L)             # None —— 列表丢了
```

### sorted 不改原列表，sort 和 reverse 改

```python
L = [9, 6, 0, 3]
a = sorted(L)
print(a, L)          # [0, 3, 6, 9] [9, 6, 0, 3] —— L 没变

b = L.sort()
print(b, L)          # None [0, 3, 6, 9] —— 原地排序，返回 None
```

### 原地修改：必须迭代下标

```python
def square_list(L):
    for i in range(len(L)):
        L[i] = L[i] ** 2

Lin = [2, 3, 4]
square_list(Lin)
print(Lin)           # [4, 9, 16]
```

### 迭代中修改列表：range 版本会正常结束

```python
L = [1, 2, 3, 4]
for i in range(len(L)):
    L.append(i)
print(L)             # [1, 2, 3, 4, 0, 1, 2, 3]
# range(len(L)) 在循环开始时就定好了，后面追加元素不会让它变长
```

### 迭代中修改列表：L = L + L

```python
L = [1, 2, 3, 4]
cnt = 0
for e in L:
    L = L + L
    cnt += 1
print('循环体执行', cnt, '次，最终长度', len(L))   # 4 次，长度 64
```

### 合并列表：+ 生成新列表，extend 原地追加

### clear() 不新建对象，L = [] 才新建

```python
L = [4, 5, 6]
before = id(L)
L.append(8)
mid = id(L)
L.clear()
after = id(L)
print(before == mid == after)   # True —— 一直是同一个对象
print(L)                        # []

L = [4, 5, 6]
j = id(L)
L = []
print(j == id(L))               # False —— 换了新对象
```

## Aliasing and Cloning

### 边迭代边删除：错误写法会漏删

```python
def remove_all_for(L, e):
    for elem in L:
        if elem == e:
            L.remove(e)

L = [1, 2, 2, 2]
remove_all_for(L, 2)
print(L)     # [1, 2] —— 漏了一个 2
```

### 正确写法：while 循环

```python
def remove_all(L, e):
    while e in L:
        L.remove(e)

L = [1, 2, 2, 2]
remove_all(L, 2)
print(L)     # [1]
```

### remove_dups：不 clone 会漏删

```python
def remove_dups_bad(L1, L2):
    for e in L1:
        if e in L2:
            L1.remove(e)

L1 = [10, 20, 30, 40]
L2 = [10, 20, 50, 60]
remove_dups_bad(L1, L2)
print(L1)    # [20, 30, 40] —— 20 被跳过了
```

### 用副本迭代就能修好

```python
def remove_dups(L1, L2):
    L1_copy = L1[:]          # 切片新建一个列表
    for e in L1_copy:
        if e in L2:
            L1.remove(e)

L1 = [10, 20, 30, 40]
L2 = [10, 20, 50, 60]
remove_dups(L1, L2)
print(L1)    # [30, 40]
```

### 别名和克隆的区别

```python
L1 = [10, 20, 30, 40]
alias = L1
clone = L1[:]
print(alias is L1)   # True —— = 只是给同一个对象起了个新名字
print(clone is L1)   # False —— 切片才真的复制
```

### 浅拷贝：顶层分开，内层仍然共享

```python
import copy
old_list = [[1, 2], [3, 4], [5, 6]]
new_list = copy.copy(old_list)
old_list.append([7, 8])    # 只动顶层
old_list[1][1] = 9         # 动的是内层
print('new:', new_list)    # [[1, 2], [3, 9], [5, 6]] —— 被内层改动波及
print('old:', old_list)    # [[1, 2], [3, 9], [5, 6], [7, 8]]
```

### 深拷贝：每一层都独立

```python
old_list = [[1, 2], [3, 4], [5, 6]]
new_list = copy.deepcopy(old_list)
old_list.append([7, 8])
old_list[1][1] = 9
print('new:', new_list)    # [[1, 2], [3, 4], [5, 6]] —— 完全没被影响
print('old:', old_list)    # [[1, 2], [3, 9], [5, 6], [7, 8]]
```

## Exceptions and Assertions

### try / except 基本用法

```python
def sum_digits(s):
    total = 0
    for char in s:
        try:
            val = int(char)
            total += val
        except:
            print("can't convert", char)
    return total

print(sum_digits('a1b2c3'))   # 打印三行提示，返回 6
```

### 分别处理不同的异常

```python
try:
    a = int('5')
    b = int('0')
    print('a/b =', a / b)
except ValueError:
    print('Could not convert to a number.')
except ZeroDivisionError:
    print("Can't divide by zero")
```

### raise 主动报错

```python
def pairwise_div(Lnum, Ldenom):
    if len(Lnum) != len(Ldenom) or not Lnum:
        raise ValueError('lists must be non-empty and of equal length')
    res = []
    for i in range(len(Lnum)):
        if Ldenom[i] == 0:
            raise ValueError('denominator is 0')
        res.append(Lnum[i] / Ldenom[i])
    return res

print(pairwise_div([4, 5, 6], [1, 2, 3]))   # [4.0, 2.5, 2.0]
```

### assert 检查前提

```python
def avg_checked(grades):
    assert len(grades) != 0, 'no grades data'
    return sum(grades) / len(grades)

print(avg_checked([80, 70, 85]))   # 78.33333333333333
# avg_checked([]) 会抛 AssertionError: no grades data
```

### 同一个问题的三种处理策略

```python
def avg_print(grades):
    try:
        return sum(grades) / len(grades)
    except ZeroDivisionError:
        print('warning: no grades data')
        return None

def avg_zero(grades):
    try:
        return sum(grades) / len(grades)
    except ZeroDivisionError:
        print('warning: no grades data')
        return 0.0

print(avg_print([]))   # 打印 warning，返回 None
print(avg_zero([]))    # 打印 warning，返回 0.0
```

### 哈希函数：键为什么必须不可变

```python
def h(s):
    # 字母序号相加，再对 16 取余
    return sum(ord(c.lower()) - ord('a') + 1 for c in s) % 16

for w in ['Ana', 'Eric', 'John', 'Kate', 'Cate']:
    print(w, '->', h(w))
# Kate 落在 5，改名成 Cate 就落到 13 —— 键一旦能改，位置就变了，值也就找不回来
```

## Reference Solutions

### 第 1 题：求和与乘积

```python
def sum_and_prod(L):
    total = 0
    prod = 1
    for e in L:
        total += e
        prod *= e
    return (total, prod)
```

### 第 2 题：生成有序列表

```python
def make_ordered_list(n):
    L = []
    for i in range(n + 1):
        L.append(i)
    return L
```

### 第 3 题：删除元素，但不改原列表

```python
def remove_elem(L, e):
    newL = []
    for elem in L:
        if elem != e:
            newL.append(elem)
    return newL

# 等价写法：return [elem for elem in L if elem != e]
```

### 第 4 题：数单词

```python
def count_words(sen):
    return len(sen.split(' '))
```

### 第 5 题：按字母序排词

```python
def sort_words(sen):
    L = sen.split(' ')
    return sorted(L)          # sorted 不改原列表，返回新列表
```

### 第 6 题：原地平方

```python
def square_list(L):
    for i in range(len(L)):
        L[i] = L[i] ** 2
    # 没有 return：目的就是改输入
```

### 第 7 题：原地删除所有等于 e 的元素

```python
def remove_all(L, e):
    while e in L:
        L.remove(e)
    # 不要写成 for elem in L，边迭代边删会漏掉元素
```

### 第 8 题：删除两个列表的共有元素

```python
def remove_dups(L1, L2):
    L1_copy = L1[:]           # 必须迭代副本
    for e in L1_copy:
        if e in L2:
            L1.remove(e)
```

### 第 9 题：逐项相除，遇 0 报错

```python
def pairwise_div(Lnum, Ldenom):
    assert len(Lnum) == len(Ldenom) and len(Lnum) != 0
    res = []
    for i in range(len(Lnum)):
        if Ldenom[i] == 0:
            raise ValueError('denominator is 0')
        res.append(Lnum[i] / Ldenom[i])
    return res
```

### 第 10 题：按名单取成绩

```python
def find_grades(grades, students):
    res = []
    for s in students:
        res.append(grades[s])
    return res

# 等价写法：return [grades[s] for s in students]
```

### 第 11 题：在字典列表里找键

```python
def find_in_L(Ld, k):
    for d in Ld:
        if k in d:
            return True
    return False
```

### 第 12 题：数 key 等于 value 的条目

```python
def count_matches(d):
    n = 0
    for k, v in d.items():
        if k == v:
            n += 1
    return n
```

### 第 13 题：嵌套字典求平均

```python
def get_average(data, what):
    all_data = []
    for stud in data.keys():
        all_data = all_data + data[stud][what]
    return sum(all_data) / len(all_data)
```

## Common Pitfalls

最后把这一讲最容易出错的地方集中过一遍。

```python
# 1. 会改输入的方法返回 None，别用赋值去接
L = [2, 1, 3]
print(L.append(5))          # None
print(L)                    # [2, 1, 3, 5]

# 2. 边迭代边删除会漏元素：用 while，或者迭代副本 L[:]

# 3. 克隆必须用切片，= 只是别名
L1 = [1, 2]
print(L1[:] is L1)          # False
alias = L1
print(alias is L1)          # True

# 4. 浅拷贝挡不住内层的改动，嵌套结构要用 copy.deepcopy

# 5. 字典的 in 检查的是键，不是值
d = {'a': 1}
print('a' in d, 1 in d)     # True False
```
